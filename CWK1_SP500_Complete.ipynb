## 1.3.1 Mean weekly return
Take the last adjusted closing price in each Friday-ending week. Keep holiday-shortened weeks and remove incomplete boundary weeks. Average the week-to-week percentage changes arithmetically.


```python
import pandas as pd

def calculate_weekly_returns(df1, shares=None):
    """Calculate weekly returns from df1 (tickers as rows, dates as columns)."""
    prices = df1 if shares is None else df1.loc[shares]

    prices = prices.T.copy()
    prices.index = pd.to_datetime(prices.index)
    prices = prices.sort_index()

    weekly_prices = prices.resample("W-FRI").last()
    return weekly_prices.pct_change(fill_method=None)
```

```python
# Keep complete Friday-ending weeks for both shares and benchmark
week = prices_for_analysis.index.to_period('W-FRI')
keep = (week.start_time >= pd.Timestamp(start)) & (week.end_time < pd.Timestamp(end))
days = prices_for_analysis.index[keep]
if len(days) < 2:
    raise ValueError('Too few trading dates in complete weeks.')
df1 = prices_for_analysis.loc[days].T
benchmark_daily = market_prices.loc[days, ['^GSPC']]
if not np.isfinite(benchmark_daily.to_numpy()).all() or benchmark_daily.le(0).any().any():
    raise ValueError('Check the benchmark prices before calculating returns.')
```

```python
# All shares weekly returns
weekly_returns = calculate_weekly_returns(df1)
```

```python
# The first complete week has no preceding complete week for comparison
market_weekly_returns = calculate_weekly_returns(benchmark_daily.T)
weekly_returns = weekly_returns.iloc[1:]
market_weekly_returns = market_weekly_returns.iloc[1:]

# Verify matching week labels and valid returns
if not weekly_returns.index.equals(market_weekly_returns.index):
    raise ValueError('Share and benchmark weeks do not match.')

if (
    not np.isfinite(weekly_returns.to_numpy()).all()
    or not np.isfinite(market_weekly_returns.to_numpy()).all()
):
    raise ValueError('Missing or infinite weekly returns found.')

# Arithmetic mean: keep decimal units for later calculations
mean_weekly_returns = weekly_returns.mean(axis=0).rename(
    'Mean_weekly_return'
)

# Save calculation evidence
weekly_returns.to_csv(OUTPUT_DIR / 'weekly_returns_2021_2024.csv')
market_weekly_returns.to_csv(OUTPUT_DIR / 'benchmark_weekly_returns.csv')
mean_weekly_returns.to_csv(OUTPUT_DIR / 'mean_weekly_returns.csv')

print('Weekly return table:', weekly_returns.shape)
print('First return week:', weekly_returns.index.min().date())
print('Last return week:', weekly_returns.index.max().date())

# Display percentages for readability
display(
    mean_weekly_returns.mul(100)
    .rename('Mean_weekly_return_pct')
    .to_frame()
    .head(10)
)
```

## 1.3.2 Weekly beta
Pair each share’s weekly returns with the benchmark on the same dates. Divide their sample covariance by the benchmark’s sample variance. The paired week count shows how much history supports each estimate.


```python
import pandas as pd
import numpy as np

def calculate_weekly_beta(weekly_returns, market_returns, min_weeks=52):
    """
    weekly_returns: DataFrame with weeks as rows and shares as columns.
    market_returns: Series of benchmark returns indexed by week.
    """
    if min_weeks < 2:
        raise ValueError("min_weeks must be at least 2.")

    results = []

    for share in weekly_returns.columns:
        paired = pd.concat(
            [weekly_returns[share].rename("share"),
             market_returns.rename("market")],
            axis=1
        ).replace([np.inf, -np.inf], np.nan).dropna()

        variance = paired["market"].var(ddof=1)
        beta = np.nan

        if len(paired) >= min_weeks and variance > 0:
            beta = paired["share"].cov(paired["market"]) / variance

        results.append({
            "Share": share,
            "Beta": beta,
            "Paired_weeks": len(paired)
        })

    return pd.DataFrame(
        results, columns=["Share", "Beta", "Paired_weeks"]
    ).set_index("Share")
```

```python
# Pass market_weekly_returns as a Series by selecting its column
beta_df = calculate_weekly_beta(
    weekly_returns,
    market_weekly_returns["^GSPC"]
)

beta_df.head()
```

```python
# Check that every share uses the full set of matched weeks
if not beta_df['Paired_weeks'].eq(len(weekly_returns)).all():
    raise ValueError('Share and benchmark coverage differs.')
if not np.isfinite(beta_df['Beta'].to_numpy()).all():
    raise ValueError('Some beta estimates need investigation.')

# Exactly two clustering features; observation counts stay separate
risk_return_features = pd.concat(
    [mean_weekly_returns, beta_df['Beta']],
    axis=1
)

risk_return_features.index.name = 'Share'

beta_df.to_csv(OUTPUT_DIR / 'weekly_beta_results.csv')
risk_return_features.to_csv(OUTPUT_DIR / 'risk_return_features.csv')

print('Risk/return feature table:', risk_return_features.shape)
print('Minimum paired weeks:', beta_df['Paired_weeks'].min())
print('Maximum paired weeks:', beta_df['Paired_weeks'].max())

# Display return percentages alongside beta and observation counts
preview = risk_return_features.join(beta_df['Paired_weeks']).copy()
preview['Mean_weekly_return_pct'] = (
    preview['Mean_weekly_return'] * 100
)

display(
    preview[
        ['Mean_weekly_return_pct', 'Beta', 'Paired_weeks']
    ].head(10)
)
```

## 1.4 Adapted RFM
The observation date is the last common trading session. Compare each day’s volume with twice the average of the previous 20 sessions. A day qualifies only when its volume is strictly above that threshold.


```python
# Calculate a separate volume threshold for each share
rfm_volume = volume_for_analysis.sort_index()
rfm_close = close_for_analysis.reindex(index=rfm_volume.index, columns=rfm_volume.columns)

if len(rfm_volume) < 80:
    raise ValueError('Need 60 sessions plus 20 preceding sessions.')
if (not np.isfinite(rfm_volume.to_numpy()).all()
        or not np.isfinite(rfm_close.to_numpy()).all()
        or rfm_volume.lt(0).any().any() or rfm_close.le(0).any().any()):
    raise ValueError('Invalid prices or volumes in the RFM inputs.')

observation_date = rfm_volume.index[-1]
latest_sessions = rfm_volume.index[-60:]

def calculate_rfm(data):
    data = data.sort_index()

    # Average of the previous 20 days, excluding the current day.
    average_volume = data["Volume"].shift(1).rolling(20).mean()

    data["HighVolume"] = data["Volume"] > 2 * average_volume

    # Only consider days with a complete volume baseline.
    valid = data.loc[average_volume.notna()].copy()

    if len(valid) < 60:
        raise ValueError("Need 60 sessions with a full volume baseline.")

    recent = valid.tail(60)
    high_volume_days = valid.loc[valid["HighVolume"]]

    # Recency in calendar days, measured at the latest data date.
    # No event in the downloaded history is recorded as missing.
    recency = (
        (valid.index[-1].date() - high_volume_days.index[-1].date()).days
        if not high_volume_days.empty
        else None
    )
    record = {
        "AsOf": valid.index[-1].date(),
        "Recency": recency,
        "Frequency": int(recent["HighVolume"].sum()),
        "Monetary": (recent["Close"] * recent["Volume"]).sum(),
        "Last_high_volume_date": high_volume_days.index.max()
    }
    data["Previous_20_mean"] = average_volume
    data["Threshold"] = 2 * average_volume
    return record, data

# Use the same study dates for every share
results = []
for ticker in candidate_tickers:
    data = pd.DataFrame({"Close": rfm_close[ticker], "Volume": rfm_volume[ticker]})
    record, _ = calculate_rfm(data)
    if record["AsOf"] != observation_date.date():
        raise ValueError(f"Check the observation date for {ticker}.")
    results.append({"Share": ticker, **record})

rfm_results = pd.DataFrame(results).set_index("Share").rename(columns={"AsOf": "Observation_date"})
rfm_results["Observation_date"] = pd.to_datetime(rfm_results["Observation_date"])
rfm_results["Recency"] = rfm_results["Recency"].astype("Int64")
print("Observation date:", observation_date.date())
print("60-session window starts:", latest_sessions[0].date())
print("Sessions in window:", len(latest_sessions))
```

## 2.1 Prepare the two feature tables
Inspect the distributions before scaling. Log-transform Monetary to reduce the influence of very large dollar values. Keep Recency in days and Frequency in event counts so a difference of one day or event has the same meaning across the range. Large Recency values can still affect distances.


```python
# Inspect skewness and flag values outside the IQR fences
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler

RANDOM_STATE = 42
N_INIT = 20
plt.style.use('seaborn-v0_8-whitegrid')

# Preserve the same shares and original units from Task 1.
risk_return_original = risk_return_features.loc[
    analysis_tickers, ['Mean_weekly_return', 'Beta']
].astype(float).copy()
rfm_original = rfm_features.loc[
    analysis_tickers, ['Recency', 'Frequency', 'Monetary']
].astype(float).copy()

assert risk_return_original.index.equals(rfm_original.index)
assert risk_return_original.index.is_unique
assert np.isfinite(risk_return_original.to_numpy()).all()
assert np.isfinite(rfm_original.to_numpy()).all()
assert rfm_original.ge(0).all().all()

# Reduce the influence of very large traded values.
data = rfm_original
features = ["Recency", "Frequency", "Monetary"]
X = data[features].copy()
X["Monetary"] = np.log1p(X["Monetary"])

risk_return_prepared = risk_return_original.copy()
rfm_prepared = X.rename(columns={"Monetary": "LogMonetary"})

diagnostic_values = risk_return_original.join(rfm_original)
diagnostic_values['LogMonetary'] = rfm_prepared['LogMonetary']
diagnostic_rows = []
outlier_flags = pd.DataFrame(index=analysis_tickers)

for feature in diagnostic_values.columns:
    values = diagnostic_values[feature]
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    flagged = (values < lower) | (values > upper)
    outlier_flags[feature] = flagged
    diagnostic_rows.append({
        'Feature': feature,
        'Count': values.count(),
        'Mean': values.mean(),
        'Median': values.median(),
        'Skewness': values.skew(),
        'Minimum': values.min(),
        'Maximum': values.max(),
        'IQR_lower': lower,
        'IQR_upper': upper,
        'IQR_flagged_shares': int(flagged.sum())
    })

feature_diagnostics = pd.DataFrame(diagnostic_rows).set_index('Feature')
feature_diagnostics.to_csv(OUTPUT_DIR / 'task2_feature_diagnostics.csv')
outlier_flags.index.name = 'Share'
outlier_flags.to_csv(OUTPUT_DIR / 'task2_outlier_flags.csv')

print('Common shares:', len(analysis_tickers))
print('IQR flags identify observations for review; no shares have been removed.')
display(feature_diagnostics)
```

## 2.2 Agglomerative clustering
Ward linkage merges the two groups that give the smallest increase in within-group squared distance. Read each tree from the bottom up. A high join indicates that quite different groups have merged. An early merge cannot be reversed.


```python
from scipy.cluster.hierarchy import linkage, dendrogram, cut_tree
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.cluster import KMeans

# Build one Ward tree per feature table
trees = {}
for name, X_scaled in matrices.items():
    # Ward linkage groups shares using Euclidean distance.
    hierarchy = linkage(X_scaled, method="ward")
    trees[name] = hierarchy
    fig, ax = plt.subplots(figsize=(10, 3.8), constrained_layout=True)
    dendrogram(hierarchy, truncate_mode='lastp', p=16, show_leaf_counts=True,
               labels=X_scaled.index.tolist(), leaf_rotation=45, color_threshold=0,
               above_threshold_color='#286471', ax=ax)
    ax.set(title=f'{name} — Ward dendrogram', ylabel='Ward linkage distance',
           xlabel='Share ticker or number of shares in a collapsed branch')
    fig.savefig(OUTPUT_DIR / ('ward_' + name.replace(' ', '_') + '.png'), dpi=140)
    plt.show()
```

### Compare two to eight groups
Compare the average silhouette score with group sizes and median profiles. A larger silhouette indicates that shares tend to be closer to their own group than their nearest competing group.


```python
# Keep the labels and median profiles for every candidate
risk_cols = ['Mean_weekly_return', 'Beta']
rfm_cols = ['Recency', 'Frequency', 'Monetary']
model_cols = {'Risk and return': risk_cols, 'RFM': rfm_cols}
labels_by_model = {}
diagnostics = []
candidate_profiles = []

def record_solution(method, name, k, labels, inertia=np.nan):
    X = matrices[name]
    groups = pd.Series(labels, index=X.index, name='Cluster')
    sizes = groups.value_counts().sort_index()
    if len(sizes) != k:
        return
    labels_by_model[method, name, k] = groups
    diagnostics.append({'Method': method, 'Features': name, 'k': k,
                        'Silhouette': silhouette_score(X, groups), 'Inertia': inertia,
                        'Smallest': sizes.min(), 'Sizes': ' / '.join(sizes.astype(str))})
    profile = feature_table[model_cols[name]].groupby(groups).median()
    profile['Shares'] = sizes
    candidate_profiles.append(profile.reset_index().assign(Method=method, Features=name, k=k))

for name, X in matrices.items():
    max_k = min(8, len(X) - 1, len(X.drop_duplicates()))
    for k in range(2, max_k + 1):
        labels = cut_tree(trees[name], n_clusters=k).ravel() + 1
        record_solution('Ward', name, k, labels)

display(pd.DataFrame(diagnostics).drop(columns='Inertia').round(3))
```

## 2.3 K-means clustering
Fit each candidate 20 times with a fixed random seed and retain the fit with the smallest within-cluster sum of squares (inertia). Inertia falls as groups are added. Use the elbow plot alongside silhouette scores and group sizes. Distances to centroids favour compact groups and can be sensitive to outliers.


```python
# Fit K-means and collect the within-cluster sums of squares
for name, X_scaled in matrices.items():
    sse = {}
    max_k = min(8, len(X_scaled) - 1, len(X_scaled.drop_duplicates()))
    for k in range(2, max_k + 1):
        kmeans = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=N_INIT)
        kmeans.fit(X_scaled)
        sse[k] = kmeans.inertia_ # SSE to closest cluster centroid
        record_solution('K-means', name, k, kmeans.labels_ + 1, sse[k])

scores = pd.DataFrame(diagnostics)
scores.to_csv(OUTPUT_DIR / 'cluster_count_comparison.csv', index=False)
all_profiles = pd.concat(candidate_profiles, ignore_index=True)
all_profiles.to_csv(OUTPUT_DIR / 'candidate_cluster_medians.csv', index=False)
display(scores[scores['Method'] == 'K-means'].round(3))

fig, axes = plt.subplots(1, 2, figsize=(10, 3.6), constrained_layout=True)
for name in matrices:
    rows = scores[(scores['Method'] == 'K-means') & (scores['Features'] == name)]
    axes[0].plot(rows['k'], rows['Inertia'], marker='o', label=name)
    axes[1].plot(rows['k'], rows['Silhouette'], marker='o', label=name)
for ax, metric in zip(axes, ['Inertia', 'Mean silhouette']):
    ax.set(xlabel='Number of clusters', ylabel=metric, xticks=range(2, 9))
    ax.legend()
fig.savefig(OUTPUT_DIR / 'kmeans_diagnostics.png', dpi=140)
plt.show()
```

## 2.4 Profile and compare the clusters
Summarise each group using the original units. Cluster numbers are identifiers within one fit. Use the adjusted Rand index (ARI) and cross-tabulations to compare which shares stay together under the two algorithms. An ARI of 1 means identical groupings, while a value near 0 indicates chance-level agreement.


```python
# Add four assignments to the share table
model_names = {('Ward', 'Risk and return'): 'Ward_risk', ('Ward', 'RFM'): 'Ward_RFM',
               ('K-means', 'Risk and return'): 'KMeans_risk', ('K-means', 'RFM'): 'KMeans_RFM'}
assignments = feature_table.copy()
for key, column in model_names.items():
    k = int(selected.loc[key, 'k'])
    assignments[column] = labels_by_model[key[0], key[1], k]
assert assignments[list(model_names.values())].notna().all().all()

# Report cluster medians in the original units
def profile_table(cluster_column):
    groups = assignments.groupby(cluster_column)
    profile = groups[feature_columns].median()
    profile.insert(0, 'Shares', groups.size())
    profile.index.name = 'Cluster'
    return profile

profiles = pd.concat({col: profile_table(col) for col in model_names.values()},
                     names=['Solution', 'Cluster'])
profiles.to_csv(OUTPUT_DIR / 'cluster_profiles_median.csv')
display(profiles.round(4))
agreement = {}
for name in matrices:
    ward = assignments[model_names['Ward', name]]
    kmeans = assignments[model_names['K-means', name]]
    agreement[name] = adjusted_rand_score(ward, kmeans)
    membership = pd.crosstab(ward, kmeans)
    membership.to_csv(OUTPUT_DIR / ('algorithm_overlap_' + name.replace(' ', '_') + '.csv'))
    print(f'{name} | Ward vs K-means ARI = {agreement[name]:.3f}')
    display(membership)
stronger = max(agreement, key=agreement.get)
print(f'The two methods agree more closely on {stronger}. '
      'The off-pattern counts in each table show where group membership depends on the algorithm.')
```

Fit a new scaler to Recency and Frequency. Run Ward again with the same shares and the cluster count selected for full RFM. Compare the memberships and the dollar-value profiles before and after removing Monetary.


```python
# Change only the feature set
rf_scaled = StandardScaler().fit_transform(rfm_original[['Recency', 'Frequency']])
rf_k = int(selected.loc[('Ward', 'RFM'), 'k'])
rf_tree = linkage(rf_scaled, method='ward', metric='euclidean')
assignments['Ward_RF'] = cut_tree(rf_tree, n_clusters=rf_k).ravel() + 1
sensitivity_ari = adjusted_rand_score(assignments['Ward_RFM'], assignments['Ward_RF'])
overlap = pd.crosstab(assignments['Ward_RFM'], assignments['Ward_RF'])
assert overlap.to_numpy().sum() == len(assignments)
print(f'Full RFM vs R and F only | k = {rf_k} | ARI = {sensitivity_ari:.3f}')
display(overlap)

value_profiles = []
for column in ['Ward_RFM', 'Ward_RF']:
    grouped = assignments.groupby(column)
    summary = grouped[rfm_cols].median().rename(columns={'Monetary':'Median_USD'})
    summary['Shares'] = grouped.size()
    summary['Q1_USD'] = grouped['Monetary'].quantile(0.25)
    summary['Q3_USD'] = grouped['Monetary'].quantile(0.75)
    value_profiles.append(summary.rename_axis('Cluster').assign(Solution=column).reset_index())
value_profiles = pd.concat(value_profiles, ignore_index=True)
display(value_profiles.round(2))
value_profiles.to_csv(OUTPUT_DIR / 'sensitivity_value_profiles.csv', index=False)
overlap.to_csv(OUTPUT_DIR / 'sensitivity_membership.csv')
assignments.to_csv(OUTPUT_DIR / 'share_features_and_assignments.csv')
```

## 3.2.3 Critical evaluation
For portfolio analysis, examine how the shares’ weekly returns move together as well as their betas. The table below compares pairwise correlations inside and across the Ward risk groups. It provides historical context for investigating a proposed portfolio. Actual portfolio risk also depends on the chosen weights.

The constituent list was collected after the study period. Using that list and requiring complete histories favours survivors and established listings. Adjusted share prices and the price-index benchmark can treat dividends differently. The split-day screen cannot verify every adjustment. Correlated features can give one activity pattern extra weight. Close multiplied by volume approximates traded value and omits bid–ask spreads and price impact. These historical groups cannot establish future returns or guarantee diversification.


```python
# Compare each distinct pair of weekly return series once
returns = weekly_returns.loc[:, assignments.index]
return_corr = returns.corr()
row, col = np.triu_indices(len(assignments), k=1)
ward_risk = assignments['Ward_risk'].to_numpy()
pair_type = np.where(ward_risk[row] == ward_risk[col], 'Within risk group', 'Across risk groups')
pairs = pd.DataFrame({'Pair_type': pair_type, 'Correlation': return_corr.to_numpy()[row, col]})
pair_summary = pairs.groupby('Pair_type')['Correlation'].agg(['count', 'median', 'mean', 'min', 'max'])
assert pair_summary['count'].sum() == len(assignments) * (len(assignments) - 1) // 2
pair_summary.to_csv(OUTPUT_DIR / 'weekly_return_pair_correlations.csv')
display(pair_summary.round(3))
```

In the saved data, median pairwise return correlation was **0.299 within** risk groups and **0.297 across** them. Buying from both groups would therefore offer little evidence of diversification on the basis of the cluster labels alone.
